##### Dirigo DB

```python
import pandas as pd
import sqlite3

from pathlib import Path

ROOT = Path.cwd()
if not (ROOT / "Dirigo").exists():
    ROOT = ROOT.parent

assert (ROOT / "Dirigo").exists(), "Không tìm thấy thư mục Dirigo"
assert(ROOT / "ACEL").exists(), "Không tìm thấy thư mục ACEL"
assert(ROOT / "DOCEL").exists(), "Không tìm thấy thư mục DOCEL"

print("Project root:", ROOT.resolve())

files = {
    "OCEL_O2O": ROOT / "Dirigo" /"OCEL_O2O.csv",
    "OCEL_PickupPlan": ROOT / "Dirigo" /"OCEL_PickupPlan.csv",
    "OCEL_Truck": ROOT / "Dirigo" /"OCEL_Truck.csv",
    "OCEL_Cargo": ROOT / "Dirigo" /"OCEL_Cargo.csv",
    "OCEL_E2O": ROOT / "Dirigo" /"OCEL_E2O.csv",
    "OCEL_Events": ROOT / "Dirigo" /"OCEL_Events.csv"
}

dataframes = {name: pd.read_csv(path) for name, path in files.items()}


sqlite_path = ROOT / "Databases_Queries" /"Dirigo.sqlite"  
conn = sqlite3.connect(sqlite_path)


for table_name, df in dataframes.items():
    df.to_sql(table_name, conn, if_exists="replace", index=False)

conn.close()

print(f"SQLite has been saved as: {sqlite_path}")
```

##### ACEL DB

```python
files = {
    "ACEL_Events": ROOT / "ACEL" /"ACEL_Events.csv",
    "ACEL_Objects": ROOT / "ACEL" /"ACEL_Objects.csv",
    "ACEL_Relations": ROOT / "ACEL" /"ACEL_Relations.csv"
}

dataframes = {name: pd.read_csv(path) for name, path in files.items()}


sqlite_path = ROOT / "Databases_Queries" /"ACEL.sqlite"  
conn = sqlite3.connect(sqlite_path)


for table_name, df in dataframes.items():
    df.to_sql(table_name, conn, if_exists="replace", index=False)


conn.close()

print(f"SQLite has been saved as: {sqlite_path}")
```

##### DOCEL DB

```python
files = {
    "DOCEL_Truck_Static": ROOT / "DOCEL" /"DOCEL_Truck_Static.csv",
    "DOCEL_Cargo_Static": ROOT / "DOCEL" /"DOCEL_Cargo_Static.csv",
    "DOCEL_Events": ROOT / "DOCEL" /"DOCEL_Events.csv",
    "DOCEL_TruckScheduledPickupWeight": ROOT / "DOCEL" /"DOCEL_TruckScheduledPickupWeight.csv",
    "DOCEL_TruckStatus": ROOT / "DOCEL" /"DOCEL_TruckStatus.csv",
    "DOCEL_TruckWeight": ROOT / "DOCEL" /"DOCEL_TruckWeight.csv",
    "DOCEL_CargoStock": ROOT / "DOCEL" /"DOCEL_CargoStock.csv",
    "DOCEL_PickupplanCargo": ROOT / "DOCEL" /"DOCEL_PickupplanCargo.csv",
    "DOCEL_PickupplanNotrs": ROOT / "DOCEL" /"DOCEL_PickupplanNotrs.csv",
    "DOCEL_PickupplanTotalWeight": ROOT / "DOCEL" /"DOCEL_PickupplanTotalWeight.csv",
    "DOCEL_TruckCargoID": ROOT / "DOCEL" /"DOCEL_TruckCargoID.csv",
    "DOCEL_TruckPickupplanID": ROOT / "DOCEL" /"DOCEL_TruckPickupplanID.csv"
}


dataframes = {name: pd.read_csv(path) for name, path in files.items()}


for name, df in dataframes.items():
    if "Static" in name:
        if "Truck ID" in df.columns:
            df.rename(columns={"Truck ID": "objectID"}, inplace=True)
        if "Cargo ID" in df.columns:
            df.rename(columns={"Cargo ID": "objectID"}, inplace=True)

    if name == "DOCEL_Events" and "Event_id" in df.columns:
        df.rename(columns={"Event_id": "eventID"}, inplace=True)


sqlite_path = ROOT / "Databases_Queries" /"DOCEL.sqlite"
conn = sqlite3.connect(sqlite_path)

for table_name, df in dataframes.items():
    df.to_sql(table_name, conn, if_exists="replace", index=False)

conn.close()
print(f"DOCEL has been saved as: {sqlite_path}")
```
